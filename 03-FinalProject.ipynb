## Research Question

Were higher per-capita $CO_2$ emissions associated with greater climate-related disaster impacts (in the form of higher mortality and reduced GDP growth) for lower-income countries than for higher-income countries between the years 2005-2023?

## Background and Prior Work

Global warming is largely caused by the accumulation of human-emitted greenhouse gases (GHG), namely, carbon dioxide ($CO_2$), which accounts for about 66% of human-emitted GHGs. <a name="cite_note-1"></a><sup>1</sup> Some sources of emission include burning fossil fuels for energy, transportation, and manufacturing. Global datasets, such as those maintained by Our World in Data, provide long-term measurements of $CO_2$ emissions, as well as their trends across countries. They show clear differences between high- and low-income nations in per-capita emission levels. <a name="cite_note-2"></a><sup>2</sup> In addition, the rise in $CO_2$ emissions result in detrimental climate change, with shifting climate patterns increasing the severity and frequency of extreme weather events such as heatwaves, floods, and storms. <a name="cite_note-3"></a><sup>3</sup> The differences between high- and low-income nations in per-capita emission levels, as well as the climate-disaster impacts experienced, raises questions about climate inequality.
\
\
There has been growing research and discussion on climate inequality, which explores the ways in which climate change may disproportionately impact different groups of people. International disaster records, such as those by EM-DAT, show that disaster-related mortality rates vary over time and across regions, with certain countries experiencing significantly higher mortality rates per 100,000 people than others. <a name="cite_note-4"></a><sup>4</sup> Despite contributing far less to GHG emissions, lower-income countries and vulnerable populations typically experience larger relative burdens of climate-driven losses, compared to wealthier nations and privileged populations. <a name="cite_note-5"></a><sup>5</sup>
\
\
We can examine these climate-related disaster impacts in the form of higher mortality and economic strain. Research from global monitoring organizations highlight that economic losses are especially exacerbated in regions with weaker infrastructure and lower capacity for disaster response. <a name="cite_note-6"></a><sup>6</sup> For example, climate-related harms can greatly reduce GDP growth in poorer regions and slow overall economic development, further contributing to the wider disparity in wealth and well-being between countries. <a name="cite_note-7"></a><sup>7</sup> Finally, multiple analyses and systematic reviews show that inequalities in capacity and access to resources can compound the harmful effects of climate change, thus worsening the outcomes for these disadvantaged groups both within and across countries. <a name="cite_note-8"></a><sup>8</sup>
\
\
These findings suggest that combining measures of $CO_2$ emissions and disaster impacts, in the form of higher mortality and reduced GDP per capita, can show us different countries' climate contribution versus vulnerability. By utilizing publicly available datasets on per-capita $CO_2$ emissions, mortality rates, and GDP per capita, we aim to quantify the relationship between human activity and climate outcomes across low- and high-income countries. This approach builds on prior work in climate change science and economic inequality research while applying data science and analysis practices and methods covered in this course.

## Section 1: The Global Wealth-Emissions Paradox

### Understanding Per-Capita CO₂ Emissions Across Income Levels

To understand climate inequality, we first examine how economic development relates to carbon emissions. This section explores the fundamental relationship between a country's wealth (measured by GDP per capita, PPP-adjusted) and its per-capita CO₂ emissions. As prior literature suggests, we expect higher income countries to emit more CO₂, the wealthier a nation, the more energy and industry it typically operates.

By establishing this wealth and emissions relationship, we lay the necessary groundwork to explore our hypothesis: if wealthier nations both emit more *and* experience fewer disaster impacts per unit of emissions, while poorer nations emit less but suffer greater impacts, this would reveal a profound structural climate inequality.

The scatter plot below uses log-log axes to account for the wide range of values, with each point representing a single country-year observation (2005–2023) colored by world region.

Import necessary libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Let us use our processed and saved data

```python
df = pd.read_csv("data/02-processed/climate_inequality_2005_2023.csv")
```

Scatter Plot - Wealth vs CO₂ Emissions

```python
fig, ax = plt.subplots(figsize=(12, 7))
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=df,
    x='gdp_per_capita_ppp_const',
    y='co2_tons_per_capita',
    hue='owid_region',
    alpha=0.6,
    edgecolor='w',
    palette='viridis',
    ax=ax
)

ax.set_xscale("log")
ax.set_yscale("log")

plt.title('The Wealth-Emissions Gap (2005–2023)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('GDP per Capita (PPP, Constant 2017 Intl $)', fontsize=12)
plt.ylabel('CO₂ Emissions (Tons per Capita)', fontsize=12)
plt.legend(title='Region', bbox_to_anchor=(1.05, 1), loc='upper left')

sns.despine()
plt.tight_layout()
plt.savefig("results/fig1_wealth_emissions_scatter.png", dpi=150, bbox_inches='tight')
plt.show()
```

### Observations from the Wealth-Emissions Relationship

The log-log scatter plot above reveals a strong, consistent positive relationship between GDP per capita and per capita CO₂ emissions across 2005–2023. Several key patterns emerge:

**Strong Regional Clustering:** Nations cluster distinctly by region and income level. European and developed Asian economies occupy the upper right of the plot (high wealth and high emissions), while African and least developed nations cluster in the lower left (reflecting both lower GDP and lower per-capita emissions).

**Emissions as a Development Marker:** The near linear relationship on log-log axes suggests that as economies develop and wealth increases, per capita emissions scale proportionally. This is likely driven by increased energy consumption, industrial activity, and transportation demand.

**Outliers of Interest:** A few nations deviate from the trend. Some wealthy nations show relatively lower emissions, potentially reflecting aggressive renewable energy adoption. Others exhibit high emissions despite moderate wealth, often characteristic of resource dependent or fossil fuel exporting economies.

This wealth and emissions relationship is a critical foundation for Section 2. If wealthier nations, confirmed here as the highest emitters, also experience the lowest disaster impacts per unit of emission, while poorer nations suffer the highest impacts per unit, this would constitute strong empirical evidence for structural climate inequality.

### Summary: What Section 1 Tells Us

Wealth and emissions are intimately connected as the wealthiest nations are the largest per capita contributors to CO₂. However, emissions alone do not tell the full story of climate impact. In Section 2, we examine whether disaster impacts are equally distributed across these income levels, or whether lower income nations bear a disproportionate burden despite their lower emissions.

## Section 2: Disaster Impact Relation to Income Bucket

### Understanding Disaster Impact Across Income Levels

This section examines the correlation between disaster impact and income level using an **inequality ratio**, defined as:

$$\text{Inequality Ratio} = \frac{\text{disaster\_deaths\_missing\_per\_100k}}{\text{co2\_tons\_per\_capita}}$$

This ratio captures how much human cost (deaths and missing persons per 100k) a country bears per ton of CO₂ it emits per capita, effectively decoupling a nation's contribution to the problem from its exposure to the consequences. We expect to observe that wealth, alongside being a cause of emissions, also serves as the primary shield against disaster mortality. That being, wealthier nations emit more but are protected by stronger infrastructure, early warning systems, and disaster response capacity.

By stratifying countries into four income quartiles using `pd.qcut` on GDP per capita, we can directly compare how the inequality ratio varies across income groups and evaluate our core hypothesis.

Import necessary library

```python
import numpy as np
```

Income Groups, Inequality Ratio, Scatter Plot

```python
# Assign income quartile groups based on GDP per capita
df['income_group'] = pd.qcut(
    df['gdp_per_capita_ppp_const'], 4,
    labels=['Low Income', 'Lower-Middle', 'Upper-Middle', 'High Income']
)

# Compute inequality ratio (avoid division by zero)
df['inequality_ratio'] = df['disaster_deaths_missing_per_100k'] / (df['co2_tons_per_capita'] + 1e-9)

# Filter to rows with actual disaster deaths for log-scale plotting
plot_df = df[df['disaster_deaths_missing_per_100k'] > 0]

plt.figure(figsize=(12, 7))
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=plot_df,
    x='gdp_per_capita_ppp_const',
    y='disaster_deaths_missing_per_100k',
    hue='owid_region',
    alpha=0.6,
    palette='viridis'
)

sns.regplot(
    data=plot_df,
    x='gdp_per_capita_ppp_const',
    y='disaster_deaths_missing_per_100k',
    scatter=False,
    color='red',
    logx=True
)

plt.xscale('log')
plt.yscale('log')
plt.title('The Resilience Gap: Wealth vs. Disaster Mortality (2005–2023)', fontsize=16, fontweight='bold')
plt.xlabel('GDP per Capita (PPP, Constant 2017 Intl $) – Log Scale', fontsize=12)
plt.ylabel('Disaster Deaths & Missing per 100k – Log Scale', fontsize=12)
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Region')
plt.tight_layout()
plt.savefig("results/fig2_resilience_gap_scatter.png", dpi=150, bbox_inches='tight')
plt.show()
```
